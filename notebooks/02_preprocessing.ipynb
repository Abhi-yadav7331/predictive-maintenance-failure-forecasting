all important libraries


```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

print("Libraries imported successfully!")
```

loading data set

```python
df = pd.read_csv("../data/raw/ai4i2020.csv")

print("Dataset shape:", df.shape)
df.head()
```

defining target and feature


```python
target = "Machine failure"

features = [
    "Type",
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]"
]

X = df[features].copy()
y = df[target].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)

X.head()
```

verifing if any leakage

```python
print("Input features:")

for column in X.columns:
    print("-", column)

print("\nTarget:", y.name)
```

splitting train/test

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

Why random_state=42?
It makes the split reproducible.
Running the notebook again gives the same train/test split.
Why stratify=y?
Because only 3.39% of our observations are failures.
Without stratification, the failure ratio could differ unnecessarily between training and testing.

verifying spliting

```python
print("Original failure rate:")
print(y.mean())

print("\nTraining failure rate:")
print(y_train.mean())

print("\nTesting failure rate:")
print(y_test.mean())
```

separating numerical and catogorical features

```python
categorical_features = [
    "Type"
]

numerical_features = [
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]"
]

print("Categorical:", categorical_features)
print("Numerical:", numerical_features)
```

building procesing pipeline

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numerical_features
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        )
    ]
)

print("Preprocessor created successfully!")
```
